```python
import random
'''
This is the board grid. It's our main data structure for this game. We can fill it only with 'X' or 'O'.
We should provide also a clear() for it that reinitializes the board (every value sets back to ''. 
'''


'''
The board should look like this, for example:

 O |   |   
-----------
   |   |   
-----------
   |   | X

'''

def visualizeBoard(board):
    gridline = "-"*11
    top_array = board[:3]
    mid_array = board[3:6]
    bot_array = board[6:9]

    print(format_row(top_array))
    print(gridline)
    print(format_row(mid_array))
    print(gridline)
    print(format_row(bot_array))

def format_row(row):
    return f" {row[0]} | {row[1]} | {row[2]} "

def change_turn(turn):
    return not turn

def mytictactoe():
    board = [' ']*9
    valid_inputs = {"nw":0, "n":1, "ne":2, "mw":3, "c":4, "me":5, "sw":6, "s":7, "se":8}
    
    availables = set(["nw", "n", "ne", "mw", "c", "me", "sw", "s", "se"])
    player_turn = False
    player_is = ''
    winner = False
    
    print('''Welcome to My Tic-Tac-Toe!\nAre you ready to rumble?\nLet's start...\n''')
    if random.randint(0, 1) == 0:
        print('''This time you are O\n
    Let me start first.\n''')
        player_is = 'O'
    else:
        print('''This time you are X\n
    Your turn to start!\n''')
        player_is = 'X'
        player_turn = change_turn(player_turn)

    while not winner:
        print("\n")
        visualizeBoard(board)
        print("\n")

        while player_turn:
            print(f'''Give an empty position in the board.\n
Possible values are: {availables}''')
            print("\n")
            player_input = input("Your turn: ")
            if not player_input in availables:
                print("Invalid input. Try again.")
            else:
                board[valid_inputs[player_input]]= player_is
                availables.remove(player_input)
            winner = check_winner(board, player_is)
            if winner:
                print("\n")
                visualizeBoard(board)
                print("\n")
                return "And the winner is "+ player_is+"! Congratulations!"
            player_turn = False
        else:
            if player_is == 'X':
                computer_is = 'O'
            else:
                computer_is = 'X'
            # Computer selects a position to play
            position = availables.pop()
            # Computer plays this position
            board[valid_inputs[position]] = computer_is
            # Checking if it's a win
            winner = check_winner(board, computer_is)
            if winner:
                return "Good Game! You'll have better luck next time :)"
            player_turn = True

        if not availables:
            return "It's a draw this time. Play again?"

def check_winner(board, player):
    # Checking for a winning row
    for i in range(0, 9, 3):
        row = board[i:i+3]
        if is_equal(row, player):
            return True
    # Checking for a winning column
    for i in range(3):
        column = [board[i], board[i+3], board[i+6]]
        if is_equal(column, player):
            return True
    # Checking for a winning diagonal
    diagonal1 = [board[0], board[4], board[8]]
    if is_equal(diagonal1,player):
        return True
    diagonal2 = [board[2], board[4], board[6]]
    if is_equal(diagonal2,player):
        return True
    return False

def is_equal(array, player):
    for elem in array:
        if elem != player:
            return False
    return True        
```
